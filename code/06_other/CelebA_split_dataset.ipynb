## Split the CelebA dataset files in three parts, training, validation and test, to make it easy create PyTorch Datasets.

Original dataset page: http://mmlab.ie.cuhk.edu.hk/projects/CelebA.html

Antonio Esteves @ UMinho, Mar 2024

```python
import os
import shutil
from   pathlib import Path

# Setup path to celebA dataset folder
data_path = Path("/home/datasets/celeba")

CELEBA_LIST_EVAL_FILE = data_path / "list_eval_partition.txt"
images_path           = data_path / "img_align_celeba"

train_dir = data_path / "train"
val_dir   = data_path / "val"
test_dir  = data_path / "test"

# If the training folder does not exist, create it
if train_dir.is_dir():
    print(f"{train_dir} directory exists.")
else:
    print(f"Did not find {train_dir} directory, creating one...")
    train_dir.mkdir(parents=True, exist_ok=True)

# If the validation folder does not exist, create it
if val_dir.is_dir():
    print(f"{val_dir} directory exists.")
else:
    print(f"Did not find {val_dir} directory, creating one...")
    val_dir.mkdir(parents=True, exist_ok=True)

# If the test folder does not exist, create it
if test_dir.is_dir():
    print(f"{test_dir} directory exists.")
else:
    print(f"Did not find {test_dir} directory, creating one...")
    test_dir.mkdir(parents=True, exist_ok=True)

# read a file with the dataset list of files where each line contains:
# FILE_NAME TAG_0_1_2
# If the TAG is 1 the file must be for training, if it is 1 must be 
# for validation, and if it is 2 must be for test.

count = 0
with open(CELEBA_LIST_EVAL_FILE, "r") as f:

    while True:
        line = f.readline()

        if not line:
            break

        fname, tag = line.strip().split(sep=' ')

        # split the dataset in 3 subsets

        if tag == '0':
            shutil.move(os.path.join(images_path, fname), train_dir)

        elif tag == '1':
            shutil.move(os.path.join(images_path, fname), val_dir)

        elif tag == '2':
            shutil.move(os.path.join(images_path, fname), test_dir)

        count += 1

print(count)

os.rmdir(images_path)
```
