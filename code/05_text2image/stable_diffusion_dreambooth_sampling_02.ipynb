```python
from   diffusers import StableDiffusionPipeline
from   PIL       import Image
```

```python
def image_grid(imgs, rows, cols):
    assert len(imgs) == rows * cols
    w, h           = imgs[0].size
    grid           = Image.new("RGB", size=(cols * w, rows * h))
    grid_w, grid_h = grid.size
    for i, img in enumerate(imgs):
        grid.paste(img, box=(i % cols * w, i // cols * h))
    return grid
```

```python
pipe     = StableDiffusionPipeline.from_pretrained('ajesteves/csestrela-dog').to("cuda")
#pipe     = StableDiffusionPipeline.from_pretrained('CompVis/stable-diffusion-v1-4').to("cuda")
```

```python
concept      = "csestrela" 
type_concept = "dog"        

prompt       = f"a photo of {concept} {type_concept} close to Eiffel"
#prompt       = f"a photo of {concept} {type_concept} walk on the grass"
#prompt       = f"a photo of {concept} {type_concept} on the beach"
#prompt       = f"a photo of {concept} {type_concept} on children's playground"
#prompt       = f"a {concept} {type_concept} plays with a soccer ball"
#prompt       = f"a {concept} {type_concept} on a park bench"

#prompt       = f"a photo of {concept} puppy close to Eiffel"
#prompt       = f"a photo of {concept} puppy walk on the grass"
#prompt       = f"a photo of {concept} puppy on the beach"
#prompt       = f"a photo of {concept} puppy on children's playground"
#prompt       = f"a {concept} puppy plays with a soccer ball"
#prompt       = f"a {concept} puppy on a park bench"

#prompt       = f"a photo of serra da estrela dog close to Eiffel"
#prompt       = f"a photo of serra da estrela dog walk on the grass"
#prompt       = f"a photo of serra da estrela dog on the beach"
#prompt       = f"a photo of serra da estrela dog on children's playground"
#prompt       = f"a serra da estrela dog plays with a soccer ball"
#prompt       = f"a serra da estrela dog on a park bench"
```

```python
# Tune the guidance to control how closely the generations follow the prompt.
# Values in the range 7-11 usually work best.
guidance_scale = 5.0

num_rows   = 1
num_cols   = 4
all_images = []

for _ in range(num_rows * num_cols):
    images = pipe(prompt, guidance_scale=guidance_scale).images
    all_images.extend(images)

image_grid(all_images, num_rows, num_cols)
```
